```python
import tskit
import numpy as np
import gaiapy as gp
import pandas as pd
from scipy.spatial.distance import pdist, squareform
from tqdm import tqdm
import time
import multiprocessing as mp
import os
```

```python
def locations(ts):
    nodes = ts.nodes()
    locs_array = []
    for node in nodes:
        if node.individual != -1:
            ind = ts.individual(node.individual)
            x = ind.location[0]
            y = ind.location[1]
            is_sample = node.is_sample()
            locs_array.append(node.id)
            locs_array.append(is_sample)
            locs_array.append(x)
            locs_array.append(y)
        
    locs_array = np.array(locs_array)
    locs = locs_array.reshape(-1, 4)
    return locs
```

```python
def get_timings(ts, out_prefix):
    sts = ts.simplify()
    ets = sts.extend_haplotypes()
    locs = locations(ets)
    sample_locations = locs[locs[:, 1] == 1][:, [0, 2, 3]]

    sts_l_start = time.time()
    sts_mpr = gp.quadratic_mpr(sts, sample_locations)
    sts_l_end = time.time()
    sts_l_run = sts_l_end - sts_l_start

    ets_l_start = time.time()
    ets_mpr = gp.quadratic_mpr(ets, sample_locations)
    ets_l_end = time.time()
    ets_l_run = ets_l_end - ets_l_start


    time_df = pd.DataFrame({
        'simp_quadratic_mpr': [sts_l_run], 
        'ets_quadratic_mpr': [ets_l_run]
    })

    
    time_df.to_csv(f"{out_prefix}_python_quadratic_gaia_times.csv",
                   mode='a', header=True, index=False)
    
    return time_df
```

```python
ts = tskit.load('tree-files/tree-S2.0-R0.trees')         
out_prefix = 'S2.0-R0_'
# sigma = 
# rep = 0
get_timings(ts, out_prefix)
```
