```python
def get_argscape_timing(ts, out_prefix):
    sts = ts.simplify()
    ets = sts.extend_haplotypes()

    sts_start = time.time()
    sts_result = argscape.infer(sts, method="gaia-quadratic")
    sts_end = time.time()
    sts_run = sts_end - sts_start

    ets_start = time.time()
    ets_result = argscape.infer(ets, method="gaia-quadratic")
    ets_end = time.time()
    ets_run = ets_end - ets_start

    time_df = pd.DataFrame({
        'sts_quadratic_time': [sts_run], 
        'ets_quadratic_time': [ets_run]
    })

    time_df.to_csv(f"{out_prefix}_argscape_quadratic_times.csv",
                   mode='a', header=True, index=False)
```

```python
ts = tskit.load('tree-files/tree-S0.2-R0.trees')         
out_prefix = 'S0.2-R0_'
get_argscape_timing(ts, out_prefix)
```
