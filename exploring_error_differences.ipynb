```python
target_node_ids = result_high_s110.index.tolist()
# for n in target_node_ids:
#     print(data[1.1][1]['node_stats']['num_parents'][n])

def count_trees_with_node(ts, node_id):
    count = 0
    id = int(node_id)
    for tree in ts.trees():
        # A node is in a tree if it is not isolated.
        # This checks if it has a parent or at least one child.
        if not tree.is_isolated(id):
            count += 1
    return count

sts_list = []
ets_list = []
for id in target_node_ids:
    sts_list.append(count_trees_with_node(sts_110, id))
    ets_list.append(count_trees_with_node(ets_110, id))
print(sts_list)
print("")
print(ets_list)
print("")

diff_list = np.arange(10)
for j in np.arange(10):
    diff_list[j] = ets_list[j] - sts_list[j]
print(diff_list)

```
