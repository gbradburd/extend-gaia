```python
def old_frame_errors_t_test(simp, ets):
    t_stat, p_val = stats.ttest_rel(simp[4], ets[4])
    error_diff = simp[4].mean() - ets[4].mean()
    print("for the tree with sigma:", simp[0].values[0], "and rep:", simp[1].values[0])
    print("the t statistic is:", t_stat)
    print("the p value is:", p_val)
    print("the mean difference (simp - ets) in error is:", error_diff)
    print(" ")
```

```python
def new_frame_errors_t_test(results, static):
    t_stat, p_val = stats.ttest_rel(results['simp_error'], results['ets_error'])
    error_diff = results['simp_error'].mean() - results['ets_error'].mean()
    print("for the tree with sigma:", static['sigma'].values[0], "and rep:", static['rep'].values[0])
    print("the t statistic is:", t_stat)
    print("the p value is:", p_val)
    print("the mean difference (simp - ets) in error is:", error_diff)
    print(" ")
```

```python
new_frame_errors_t_test(S1_1_R2_results, S1_1_R2_static)
new_frame_errors_t_test(S0_5_R3_results, S0_5_R3_static)
new_frame_errors_t_test(S0_8_R0_results, S0_8_R0_static)
old_frame_errors_t_test(S0_5_R5_simp, S0_5_R5_ets)
old_frame_errors_t_test(S0_5_R4_simp, S0_5_R4_ets)
old_frame_errors_t_test(S0_2_R0_simp, S0_2_R0_ets)
```
