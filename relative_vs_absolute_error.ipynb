```python
def rel_error_by_time(df, static):
    fig = go.Figure()

    sigma = static['sigma'].values[0]
    rep = static['rep'].values[0]

    fig.add_trace(go.Scatter(x=df['node_time'], y = df['ets_rel_error'], mode='markers', name='Extended'))
    fig.add_trace(go.Scatter(x=df['node_time'], y = df['sts_rel_errpr'], mode='markers', name='Simplified'))

    fig.update_layout(title=f"S{sigma} R{rep} Simplified Vs Extended Relative Error by Time", xaxis_title='Time', yaxis_title='Absolute Error')
    fig.show()
```

```python
def abs_error_by_time(df, static):
    fig = go.Figure()

    sigma = static['sigma'].values[0]
    rep = static['rep'].values[0]

    fig.add_trace(go.Scatter(x=df['node_time'], y = df['ets_abs_error'], mode='markers', name='Extended'))
    fig.add_trace(go.Scatter(x=df['node_time'], y = df['sts_abs_error'], mode='markers', name='Simplified'))

    fig.update_layout(title=f"S{sigma} R{rep} Simplified Vs Extended Absolute Error by Time", xaxis_title='Time', yaxis_title='Absolute Error')
    fig.show()
```
